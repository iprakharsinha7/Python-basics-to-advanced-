##### Write a program to check if name is starting with M and age is above 18, then allowed entry

```python
name = input("Please enter your name:")
age = int(input("Please enter your age:"))

if name[0].upper() == 'M' and age >= 18 :
  print("Allowed")

elif name[0].upper() == 'M' and age < 18 :
  print("Not Allowed because age is less than 18")

elif name[0].upper() != 'M' and age >= 18 :
  print("Not Allowed because name does not starts with M")

elif name[0].upper() != 'M' and age < 18 :
  print("Not allowed because of name and age")
```

```python
name = input("Please enter your name:")

if name[0].upper() == 'M' :
  print("Name starts with M")
  age = int(input("Please enter your age:"))
  if age >= 18 :
    print("Allowed")

  else :
    print("Not Allowed")

else :
  print("Name does not starts with M")

```
