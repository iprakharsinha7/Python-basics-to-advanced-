***How to create a DataFrame***

```python
di1 = {'Name': ['Prakhar','Manisha','Prabhav','Kunal','Sandeep'],
       'Age': [30,29,28,25,27],
       'Gender': ['M','F','M','M','M']}
```

```python
df = pd.DataFrame(di1)

df
```

```python
#one can create it directly without di

df = pd.DataFrame({'Name': ['Prakhar','Manisha','Prabhav','Kunal','Sandeep'],
       'Age': [30,29,28,25,27],
       'Gender': ['M','F','M','M','M']})

df
```

```python
#create a dataframe using array

import numpy as np

arr1 = np.array(['Prakhar','Manisha','Prabhav','Kunal','Sandeep'])
arr2 = np.array([30,29,28,25,27])

df = pd.DataFrame({'Name': arr1,
                  'Age': arr2})

df
```
