##### Reshape()

```python
arr1d = np.array([1,2,3,4,5,6])
arr1d
```

```python
arr1d.shape
```

```python
arr2d = arr1d.reshape(2,3) #(2,3) (1,6)
arr2d
```

```python
arr2d.shape
```

```python
arr3d = arr1d.reshape(2,3,1)
arr3d
```

```python
arr3d.shape
```

```python
arr3d.ndim
```

```python
arr3d.size
```
