```python
# convert numbers into odd and even

li = [1,2,3,4,5,6]

odd_even = [] # empty list required

for i in li:

  if i%2==0:
    odd_even.append('Even')

  else:
    odd_even.append('Odd')

print(odd_even)
```

```python
# 1. List Comprehension (when you need to solve a problem within list bracket.. here "res")

li = [1,2,3,4,5,6]

res = [i**2 for i in li]
print(res)
```

```python
#identify odd or even

li = [1,2,3,4,5,6]

res = ["Even" if i%2==0 else "Odd" for i in li]

print(res)
```

```python
# 2. Dictionary Comprehension

li = [1,2,3,4,5,6]

di = {i:i for i in li}
di
```

```python
# Calculate the square of dictionary

li = [1,2,3,4,5,6]

di = {i:i**2 for i in li}
di
```

```python
#set to dictionary

s1 = ['Prakhar', 'Manisha', 'Sandeep', 'Kunal', 'Prabhav']

di = {i:len(i) for i in s1}  #output for_loop
di
```

#### Check the grade of a student with user defined function

```python
def grade(score):

  if score<33:
    print("Fail")

  elif score<60:
    print("B")

  elif score<80:
    print("A")

  else:
    print("A+")
```

```python
grade(99)
```

```python
grade(30)
```

```python
grade(61)
```

```python
grade(59)
```
